```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

#Fortwsh dedomenwn
url_diabetes = "https://raw.githubusercontent.com/npradaschnor/Pima-Indians-Diabetes-Dataset/master/diabetes.csv"
df_diabetes = pd.read_csv(url_diabetes)

print("Πληροφορίες Dataset Διαβήτη")
print(df_diabetes.info())

#Dhmioyrgia pinakwn X kai Y
X=df_diabetes.drop('Outcome',axis=1).values
y=df_diabetes['Outcome'].values

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.30,random_state=0)
print("\n Επαλήθευση Διαχωρισμού")
print(f"Συνολικό πλήθος Χ: {X.shape[0]}")
print(f"Πλήθος X_train: {X_train.shape[0]} ({len(X_train)/len(X)*100:.1f}%)")
print(f"Πλήθος X_test: {X_test.shape[0]} ({len(X_test)/len(X)*100:.1f})")

#Kanonikopoihsh dedomenwn [0,1]
scaler=MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

classifier=LogisticRegression(random_state=0)
classifier.fit(X_train_scaled, y_train)
train_preds=classifier.predict(X_train_scaled)
test_preds=classifier.predict(X_test_scaled)

print("\n Αποτελέσματα Ακριβείας")
print(f"Accuracy sto Train set: {accuracy_score(y_train,train_preds):.4f}")
print(f"Accuracy sto Test set:{accuracy_score(y_test,test_preds):.4f}")
feature_names=df_diabetes.columns.drop('Outcome')
importances=np.abs(classifier.coef_[0])

indices=np.argsort(importances)[::-1]
plt.figure(figsize=(10,6))
plt.title("Σημαντικότητα Χαρακτηριστικών")
plt.bar(range(X.shape[1]),importances[indices],align="center", color="skyblue")
plt.xticks(range(X.shape[1]),[feature_names[i] for i in indices], rotation=45)
plt.ylabel("Απόλυτο βάρος Συντελεστή")
plt.tight_layout()
plt.show()

conf_train=confusion_matrix(y_train,train_preds)
conf_test= confusion_matrix(y_test,test_preds)
print("\n Λανθεσμένες προβλέψεις")
print(f"Στο train set: {conf_train[1,0]} περιπτώσεις")
print(f"Στο test set: {conf_test[1,0]} περιπτώσεις")
```
